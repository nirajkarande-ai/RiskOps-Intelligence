```python
# Phase-3 — Investigation Engine
```

```python
# Step 1 — Import Libraries
```

```python
import pandas as pd
import numpy as np
```

```python
# Step 2 — Load Alerts Dataset
```

```python
alerts = pd.read_csv("alerts_table (1).csv")

print("Alerts Shape:", alerts.shape)

alerts.head()
```

```python
# Step 3 — Simulate Investigator Pool
```

```python
num_investigators = 25

investigators = [f"INV_{i}" for i in range(1, num_investigators + 1)]
```

```python
# Step 4 — Assign Investigator to Each Alert
```

```python
alerts["Investigator_ID"] = np.random.choice(
    investigators,
    size=len(alerts)
)
```

```python
# Step 5 — Generate Investigation Start Time
```

```python
alerts["Alert_Date"] = pd.to_datetime(alerts["Alert_Date"])

alerts["Investigation_Start"] = (
    alerts["Alert_Date"] +
    pd.to_timedelta(np.random.randint(5, 120, size=len(alerts)), unit="m")
)
```

```python
# Step 6 — Simulate Investigation Duration
```

```python
alerts["Resolution_Time_Minutes"] = np.random.randint(
    10,
    180,
    size=len(alerts)
)
```

```python
alerts["Investigation_End"] = (
    alerts["Investigation_Start"] +
    pd.to_timedelta(alerts["Resolution_Time_Minutes"], unit="m")
)
```

```python
# Step 7 — Determine Investigation Outcome
```

```python
def determine_outcome(tier):

    if tier == "High":
        return np.random.choice(["Fraud","False Positive"], p=[0.9,0.1])

    elif tier == "Medium":
        return np.random.choice(["Fraud","False Positive"], p=[0.6,0.4])

    else:
        return "False Positive"
```

```python
alerts["Outcome"] = alerts["Risk_Tier"].apply(determine_outcome)
```

```python
# Step 8 — Update Investigation Status
```

```python
alerts["Investigation_Status"] = "Completed"
```

```python
# Step 9 — Create Final Investigations Table
```

```python
investigations = alerts[[
"Alert_ID",
"Investigator_ID",
"Investigation_Start",
"Investigation_End",
"Resolution_Time_Minutes",
"Outcome"
]]
```

```python
investigations.head()
```

```python
# Step 10 — Save Investigations Dataset
```

```python
investigations.to_csv(
    "investigations_table.csv",
    index=False
)
```
